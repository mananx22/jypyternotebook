```python
## updated file with changs in train test split sequences
import pandas as pd
messages=pd.read_csv('/home/manan/jypyternotebook/NLP/spam.csv',  skiprows=[0], usecols=[0, 1], names=["label","message"])
```

```python
messages
```

```python
## Data cleaning and Preprocessing
import re
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
corpus=[]
for i in range(0,len(messages)):
    review=re.sub('[^a-zA-Z]',' ', messages['message'][i])
    review=review.lower()
    review=review.split()
    review = [ps.stem(word) for word in review if not word in stopwords.words('english')]
    review=' '.join(review)
    corpus.append(review)
```

```python
corpus
```

## Create BOW

```python
## Output features
Y=pd.get_dummies(messages['label']).astype(int)
Y=Y.iloc[:, 1].values
Y
```

```python
## train test split
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(corpus,Y,test_size=0.20)
```

```python
## Create Bag of Words model
from sklearn.feature_extraction.text import CountVectorizer
```

```python
## for binary BOW, enable binary=true
cv=CountVectorizer(max_features=2500, ngram_range=(1,2))
X_train=cv.fit_transform(X_train).toarray()
X_train
```

```python
X_test=cv.transform(X_test).toarray()
X_test
```

```python
import numpy as np
np.set_printoptions(edgeitems=30, linewidth=100000, formatter=dict(float=lambda x: "%.3g" % x))
```

```python
X_train
```

```python
cv.vocabulary_
```

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
Y_train
```

```python
spam_detect_model=MultinomialNB().fit(X_train,Y_train)
```

```python
Y.ndim # to check if it is numpy array ndim == 1
```

```python
Y_pred=spam_detect_model.predict(X_test)
```

```python
from sklearn.metrics import classification_report, accuracy_score
```

```python
accuracy_score(Y_test,Y_pred)
```

```python
print(classification_report(Y_test,Y_pred))
```

## Create TF-Idf Model

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tv=TfidfVectorizer(max_features=2500, ngram_range=(1,2))
```

```python
## train test split
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(corpus,Y,test_size=0.20)
```

```python
X_train=tv.fit_transform(X_train).toarray()
X_test=tv.transform(X_test).toarray()
X_train
```

```python
tv.vocabulary_
```

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
spam_tfidf=MultinomialNB().fit(X_train,Y_train)
```

```python
## pred
y_pred=spam_tfidf.predict(X_test)
```

```python
score=accuracy_score(Y_test,y_pred)
```

```python
print(score)
```

```python
print(classification_report(Y_test,y_pred))
```
