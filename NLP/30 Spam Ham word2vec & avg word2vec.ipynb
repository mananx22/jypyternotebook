```python
from nltk.stem import WordNetLemmatizer
lemmitizer=WordNetLemmatizer()
```

```python
## Data cleaning and Preprocessing
import re
import nltk
nltk.download('stopwords')
import re
corpus=[]
for i in range(0,len(messages)):
    review=re.sub('[^a-zA-Z]',' ', messages['message'][i])
# Uses regex to replace all non-alphabetic characters (numbers, symbols, etc.) with a space.
# Example: "Hello! 123 World." → "Hello World "
    review=review.lower()
    review=review.split()
    review = [lemmitizer.lemmatize(word) for word in review]
    review=' '.join(review)
    corpus.append(review)    
# Example Input → Output
# Input Message (raw):
# "Hey! I'm learning NLP. It's awesome! 123."
# Processed Output in corpus:
# "hey i m learning nlp it s awesome"
```

```python
len(corpus)
```

```python
from nltk import sent_tokenize
from gensim.utils import simple_preprocess
```

```python
words=[]
```

```python
# aletnative to using the above method to make corpus lowercae and splitting
for sent in corpus:
    sent_token=sent_tokenize(sent)
    for sent in sent_token:
        words.append(simple_preprocess(sent))
```

```python
import gensim
## lets train word2vec from scratch
model=gensim.models.Word2Vec(words)
```

## Average word2vec

```python
!pip install tqdm
import numpy as np
def avg_word2vec(doc):
    return np.mean([model.wv[word] for word in doc if word in model.wv.index_to_key],axis=0)
```

```python
from tqdm import tqdm
```

```python
# apply for the entire sentences
X=[]
for i in tqdm(range(len(words))):
    X.append(avg_word2vec(words[i]))
```
