IMPORTING Pandas and MATPLOTLIB for generating graph

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```python
# reading dataset
df=pd.read_csv('weight-height.csv')
```

```python
#showing dataset
df=df.drop(['Gender'],axis=1)
df.head()
```

Plotting our DATA

```python
# scatter Plot
plt.scatter(df['Weight'],df['Height'])
plt.xlabel('weight')
plt.ylabel('height')
```

### We see weight is directly proportional to height. 
To check whether the correlation between height and weight is positive or negative we will do following

```python
## correlation
df.corr()
```

```python
# We can also use one more library for visualization
import seaborn as sns
sns.pairplot(df)
```

Dividing our features into Independent and Dependent features

```python
X=df[['Weight']]  # Always make sure your independent feature is dataframe or 2d array or have columns
Y=df['Height'] # your dependent feature can be series or 1d 
X.head()
```

```python
# to see size 
np.array(X).shape
```

Train TEST split

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.01,random_state=42)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train) # fit_transform calculate mean and standard deviation for provided data. only required for train.
```

```python
X_test=scaler.transform(X_test) #not required for test
```

```python
X_test
```

# Apply simple linear regression

```python
from sklearn.linear_model import LinearRegression
```

```python
regression = LinearRegression(n_jobs=-1)
```

```python
regression.fit(X_train,Y_train)
```

```python
print("coeficient or slope:", regression.coef_)
print("Intercept:", regression.intercept_)
```

```python
# Plot best fit line wrt training  data
plt.scatter(X_train,Y_train)
plt.plot(X_train,regression.predict(X_train))
```

```python
# prediction for test data
Y_pred = regression.predict(X_test)
```

```python
# performance Metrics
from sklearn.metrics import mean_squared_error,mean_absolute_error
```

```python
mse=mean_squared_error(Y_test,Y_pred)
mae=mean_absolute_error(Y_test,Y_pred)
rmse=np.sqrt(mse)
print("mse",mse," mae:",mae," rmse:",rmse)
```

```python
from sklearn.metrics import r2_score
```

```python
score=r2_score(Y_test,Y_pred)
print(score)
```

# Adjusted R square


```python
1 - (1-score)*(len(Y_test)-1)/(len(Y_test)-X_test.shape[1]-1)
```
