```python
import seaborn as sns
```

```python
df=sns.load_dataset("tips")
df.head()
```

```python
## now try to predict using naive byes by converting categorical into numerical
```

```python
#independent and dependent feature
X=df[['sex']]
Y=df['time']
Y
```

```python
#before moving on to perform one hot encoding for binary we first perform split of dataset
```

```python
# Split the dataset into traintest split
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.25,random_state=0)
# X_train.head()
Y_train.head()
```

```python
# applying label encoding on traning data
from sklearn.preprocessing import LabelEncoder
le1=LabelEncoder()
X_train['sex']=le1.fit_transform(X_train['sex'])
le2=LabelEncoder()
Y_train=le2.fit_transform(Y_train)
Y_train
```

```python
X_train
```

```python
# applying same on testlabel encoding
X_test['sex']=le1.fit_transform(X_test['sex'])
Y_test=le2.fit_transform(Y_test)
Y_test
```

```python
from sklearn.naive_bayes import  GaussianNB
gnb=GaussianNB()
gnb.fit(X_train,Y_train)
```

```python
Y_Pred=gnb.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
print(accuracy_score(Y_Pred,Y_test))
print("--------------")
print(confusion_matrix(Y_Pred,Y_test))
print("--------------")
print(classification_report(Y_Pred,Y_test))
```
