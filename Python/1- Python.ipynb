### Map Function
A map function applies a given function to all items in an input list (array) and output as map object or iterator. THis is particularly useful to transform data in  a list comprehensively.

```python
def square(x):
    return x*x

numbers = [3,4,5,6,7,8]

map(square,numbers)
```

```python
# to view map
list(map(square,numbers))
```

```python
# lambda function with map
numbers = [4,3,6,7,21]
k = lambda x:x*x 
list(map(k,numbers))
```

```python
# example 2  :map with  multiple iterables
num1 = [2,3,4]
num2 = [4,6,7]
# to add
k = lambda x,y:x+y
list(map(k,num1,num2))
```

```python
# example 3: map to convert list of strings to integers
val = ["4","6","7"]
list(map(int,val))
```

```python
# example 4 : appply map on dictionary
dic = [
    {"name": "manan", "city":"jodhpur"},
    {"name": "kratika", "city":"bikaner"}
]

def get_name(per):
    return per["city"]

list(map(get_name,dic))
```

### filter()
filter function constructs an iterator from the iterable elements for which a function returns true. it is used to filter out items from a list( or any other iterable) based on condition.

```python
lister = [3,4,5,3,2,4,5,7,9,2,10]
even = lambda x:x%2==0
greater = lambda x:x>5
# list(filter(even,lister))
# list(filter(greater,lister))
list(filter(lambda x:x>5 and x%2==0,lister))
```
