# Linear Search
Linear Search is a simple algo, in which each element of the list is compared against the target and return the index if found, return -1 if not found.

```python
def linear_search(list,int):
  inside = 0 
  for i in range(0,len(list)):
    if(list[i] == int):
      inside = i
      break
    else:
      inside = -1
  return inside

my_list = [25,34,52,67,89,4,5,6,8]
target = 89
result = linear_search(my_list,target)
print(result)
```
