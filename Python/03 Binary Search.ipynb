# Binary Search
Here, Array is Always **Sorted**.
* Either Increasing
* Either Decreasing

```python
def binary(list,int):
    inside_value = 0
    start_index = 0
    end_index = len(list)-1
    while(start_index <= end_index):
        mid = (end_index + start_index)//2
        print("start" + str(start_index))
        print("end"+ str(end_index))
        print("mid" + str(mid) )
        if(list[mid]==int):
            return mid
        elif(list[mid] < int):
            start_index = mid + 1
        else:
            end_index = mid -1
        print("Iteration done ++++")
    return -1


# we will start with sorted array
array = [10,23,35,45,50,70,85]
target = 50
result = binary(array,target)
print(result)
```
