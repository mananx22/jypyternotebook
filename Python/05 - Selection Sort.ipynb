In selection sort, 
1) first we identify the minimum value
2) next we replace minimum value with the leftest value.
3) cycle repeats.

```python
def selection_sort(list):
    for i in range(len(list)):
        min = list[i]
        for j in range(i+1, len(list)):
            if (list[i] > list[j]):
                k = list[i]
                list[i] = list[j]
                min = list[j]
                list[j] = k
                print(str(min) + "================")
                print(list)
                
    return list

print(selection_sort([12,25,11,34,90,22]))
```
