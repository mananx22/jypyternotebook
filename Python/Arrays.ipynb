# **Arrays**
Arrays are collection of elements stored within continuous memory location. 

Arrays are:
* Homogenous
* fixed size
* allow for efficient access and manipulation of data

In pyhton we use list instead of arrays as they provide much more flexibility. however, python provide ability to use arrays via libraries such as numpy.




Arrays are not provided by python by default, we need to import them. List & Dictionaries are provided by default.

```python
import array
a = array.array('i', [1, 2, 3, 4, 5])
print(a)
print(type(a))
```

Update a value in array

```python
a[0] = 100
print(a)
```
