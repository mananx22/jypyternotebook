Bubble sort algo
Plan of action:
Bubble Sort = “Keep swapping neighbors”

You walk along the line and look at two kids at a time:

If the kid on the left has a bigger number than the kid on the right
👉 you say: “Oops! Swap places!”

You keep doing this again and again.

```python
def bubble_sort(list):
        start = 0
        end = len(list)-1
        for start in range (0,end):
            for start in range (0,end) :
                if(list[start]<list[start+1]):
                    # print("value is" + str(list[start]) + str(k) + str(list[start+1]))
                    print("sorted Iteration" + str(start) + str(start+1) + "----" + str(list))
                else:
                    k = list[start]
                   
                    # print("value of k is" + str(list[start]) + str(k) + str(list[start+1]))
                    
                    list[start] = list[start+1]
                    list[start+1] = k
                    print("Updated Iteration" + str(start) + str(start+1) + "----" + str(list))
                    # return list
            end = end -1
            # return
        return list




given_list = [12,25,11,34,90,22]
returned_list = bubble_sort(given_list)
print(returned_list)
```
