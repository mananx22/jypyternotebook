```python
def bubble_sort(list):
        start = 0
        end = len(list)-1
        for start in range (0,end) :
            if(list[start]<list[start+1]):
                print("sorted Iteration" + str(start) + str(start+1) )
            else:
                k = list[start+1]
                list[start] = list[start+1]
                list[start+1] = k
                print("Updated Iteration" + str(start) + str(start+1) )
        return list



given_list = [12,25,11,34,90,22]
returned_list = bubble_sort(given_list)
print(returned_list)
```
