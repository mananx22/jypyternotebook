### Inheritance in python
inheritance is a fundamental property of python class that allows one class to fetch properties from other python classes.

````python
# Inheritance Example
#Parent class 1
class car:
    def __init__(self,windows,doors,enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype

    def drive(self):
        print(f"the person will drive car with {self.enginetype} engine")   

#Parent class 2
class place:
    def __init__(self,country):
        self.country = country 

````

````python
# inheriting class/ child class
class tesla(car,place):
    def __init__(self, windows, doors, enginetype,automatic,country):
        car.__init__(self,windows, doors, enginetype)
        place.__init__(self,country)
        self.automatic=automatic
    def drive(self):
        print (f"country is = {self.country}")
        return car.drive(self)
    
car2 = tesla(2,3,"v8","yes","ind")
````

````python
car2.drive()
````

### Polymorphisom 
polymorphism allow different classes to be treated as common superclass. it allows a way to perform single action in multiple ways. Polymorphism is achieved through method overriding and interfaces.

````python
class area:
    def __init__(self,num):
        self.num = num

    def area_function(self):
        return self.num*self.num
    
class rectangle(area):
    def __init__(self,num,bre):
        self.num = num
        self.bre = bre
    
    def area_function(self):
        return self.num*self.bre
    
class circle(area):
    def __init__(self,num):
        self.num = num

    def area_function(self):
        return self.num*self.num*3.14
    


````

````python
square = area(4)
square.area_function()

````

````python
rectangle = rectangle(3,4)
rectangle.area_function()
````

````python

circle = circle(4)
circle.area_function()
````

### Polymorphism using abstract classes

Alright — this is a **core OOP concept**, but I’ll make it very simple 👇

---

# 🧠 Big idea

👉 **Polymorphism = same action, different behavior**

👉 **Abstract class = rule book**

---

# 🧃 ELI5 analogy

Think:

* Abstract class = **remote control template 📺**
* Child classes = **TV, AC, Speaker**

👉 All must have a `turn_on()` button
But each behaves differently

---

# ✅ Step-by-step example

---

## 1️⃣ Abstract class (rule book)

```python
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass
```

👉 This says:

> “Every animal MUST have `sound()` method”

---

## 2️⃣ Child classes (different behaviors)

```python
class Dog(Animal):
    def sound(self):
        return "Bark"

class Cat(Animal):
    def sound(self):
        return "Meow"
```

---

## 3️⃣ Polymorphism in action

```python
animals = [Dog(), Cat()]

for a in animals:
    print(a.sound())
```

👉 Output:

```
Bark
Meow
```

---

# 🧠 What just happened?

👉 Same method:

```python
sound()
```

👉 Different outputs:

* Dog → Bark
* Cat → Meow

👉 That’s **polymorphism**

---

# ⚠️ Important rule

You CANNOT do this:

```python
a = Animal() ❌
```

👉 Because it’s abstract (incomplete)

---

# 🧃 Super simple version

| Concept        | Meaning                           |
| -------------- | --------------------------------- |
| Abstract class | “You MUST implement this”         |
| Polymorphism   | “Same function, different result” |

---

# 🎯 One-line takeaway

👉 Abstract class defines rules
👉 Polymorphism follows those rules differently

---

# 💪 Why this matters (real-world)

* Ensures consistent API
* Used in ML pipelines, frameworks, large apps
