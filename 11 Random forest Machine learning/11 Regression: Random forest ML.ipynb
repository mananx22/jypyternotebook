### Split the dataset in X & Y

```python
from sklearn.model_selection import train_test_split
X = df.drop('selling_price', axis=1)
Y = df['selling_price']
```

#### Apply label encoding
this is done because we see model column hv 120 repetitive elements. instead of apply ohe we do label encoding.
label encoder apply labels for eg: [1,1,3,6] -> [0,0,1,2]

```python
from sklearn.preprocessing import LabelEncoder
ls = LabelEncoder()
X['model'] = ls.fit_transform(X['model'])
```

```python
#as we see our X is having labels as model
X.head()
```

#### OHE will create 119 extra columns thats why we went with labels approach.

```python
X.info()
```

```python
# alternatively num_features = [feature for feature in X.columns if X[feature].dtype != "O" ]
cat_feature = X.select_dtypes('object')
num_feature = X.select_dtypes(exclude='object')
```

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer

ohe = OneHotEncoder(drop='first')
ss = StandardScaler()

preprocessor = ColumnTransformer(
    [
        ("ohe", ohe, cat_feature.columns),
        ("StandardScaler", ss, num_feature.columns)
    ], remainder= "passthrough"
)
```

```python
preprocessor
# cat_feature.head()
# num_feature.info()
```

```python
pd.DataFrame(X)
```

```python
X = preprocessor.fit_transform(X)
pd.DataFrame(X)
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y,random_state=42,test_size=0.2)
```

### Model training and model selection

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import root_mean_squared_error,mean_absolute_error,mean_squared_error,r2_score
```

```python
# create a function to evaluate metrics
def evaluatin(true,pred):
    rmse = root_mean_squared_error(true,pred)
    mae = mean_absolute_error(true,pred)
    mse = mean_squared_error(true,pred)
    r2_square = r2_score(true, pred)
    return rmse,mae,mse, r2_square
```

```python
# Creating models
models = {
    "RandomForestRegressor": RandomForestRegressor(),
    "DecisionTreeRegressor": DecisionTreeRegressor(),
    "LinearRegression": LinearRegression(),
    "Ridge": Ridge(),
    "Lasso": Lasso(),
    "KNeighborsRegressor": KNeighborsRegressor()
}
```

```python
list(models.values())[1]
```

```python
for i in range(len(models)):
    model = list(models.values())[i]

    model.fit(X_train,Y_train) # train model
    # test models
    X_train_pred = model.predict(X_train)
    X_test_pred = model.predict(X_test)
    
    rmse,mae,mse, r2_square = evaluatin(Y_train, X_train_pred)

    print( "selected model is:", model)
    print("values for Train is:")
    print ( "root_mean_squared_error: ", format(rmse))
    print ( "mean absolute error:", format(mae))
    print("mean squared error", format(mse))
    print( "r2 score", format(r2_square))

    rmse,mae,mse, r2_square = evaluatin(Y_test, X_test_pred)

    print("values for Test is:")
    print ("root_mean_squared_error: ", format(rmse))
    print ( "mean absolute error:", format(mae))
    print("mean squared error", format(mse))
    print( "r2 score", format(r2_square))
    print('\n')
    print('\n')
```

### HyperParameter Tuning
we observed best performance was given by k nearest and random forest regression

```python

kn_params = {"n_neighbors": [2,3,10,20,30,40]}
rf_params = {"max_depth": [5, 8, 15, None, 10],
             "max_features": [5, 7, None, 8],
             "min_samples_split": [2, 8, 15, 20],
             "n_estimators": [100, 200, 500, 1000]
            }
```

```python
random_cv_model = [
    ("knn", KNeighborsRegressor(), kn_params),
    ("rf", RandomForestRegressor(), rf_params)
    ]


```

```python
from sklearn.model_selection import RandomizedSearchCV
models = {}
for name,model,params in random_cv_model:
    rando = RandomizedSearchCV(estimator=model,
                               param_distributions=params,
                               n_iter=100,
                               cv=3,
                               n_jobs=-1)
    rando.fit(X_train,Y_train)
    models[name] = rando.best_estimator_, rando.best_score_

for name in models:
    print(f"---------------- Best Params for {name} -------------------")
    print(models[name])
```
