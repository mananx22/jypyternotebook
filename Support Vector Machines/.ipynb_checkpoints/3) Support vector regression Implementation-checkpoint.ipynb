```python
## since the above three categories are binary
# we do FEATURE ENCODING by applying label encoding with 0s and 1s
# for 'day'we apply one-hot encoding 
df.columns
```

```python
#independent and dependent feature
X=df[['tip', 'sex', 'smoker', 'day', 'time', 'size']]
Y=df['total_bill']
Y
```

```python
# Split the dataset into traintest split
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.25,random_state=10)
X_train.head()
```

```python
# we should split our dataset first before procedding for feature encoding as there will always be a possibility of data leakage i.e the models should
# not have any information about the test data.
```

```python
# applying label encoding
from sklearn.preprocessing import LabelEncoder
le1=LabelEncoder()
X_train['sex']=le1.fit_transform(X_train['sex'])
le2=LabelEncoder()
X_train['smoker']=le2.fit_transform(X_train['smoker'])
le3=LabelEncoder()
X_train['time']=le3.fit_transform(X_train['time'])
X_train
```

```python
# applying same on testlabel encoding
from sklearn.preprocessing import LabelEncoder
le1=LabelEncoder()
X_test['sex']=le1.transform(X_test['sex'])
le2=LabelEncoder()
X_test['smoker']=le2.transform(X_test['smoker'])
le3=LabelEncoder()
X_test['time']=le3.transform(X_test['time'])
X_test
```
