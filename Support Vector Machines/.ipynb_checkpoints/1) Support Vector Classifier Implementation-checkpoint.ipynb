```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns 
%matplotlib inline
```

```python
## Let's create synthetic DataPoints
from sklearn.datasets import make_classification
```

```python
X,Y=make_classification(n_samples=1000,n_features=2,n_classes=2,n_clusters_per_class=2,n_redundant=0,n_informative=2)
```

```python
X
```

```python
pd.DataFrame(X)
```

```python
Y
```

```python
sns.scatterplot(x=pd.DataFrame(X)[0],y=pd.DataFrame(X)[1],hue=Y)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.25,random_state=42)
```

```python
from sklearn.svm import SVC
```

```python
svc=SVC(kernel='linear')
```

```python
svc.fit(X_train,Y_train)
```

```python
## Predction
Y_pred=svc.predict(X_test)
```

```python
from sklearn.metrics import classification_report,confusion_matrix
print(classification_report(Y_test,Y_pred))
print(confusion_matrix(Y_test,Y_pred))
```

```python
## Now we will improve above accuracy using rbf kernel
rbf=SVC(kernel='rbf')
```

```python
rbf.fit(X_train,Y_train)
```

```python
## Predction
Y_pred1=rbf.predict(X_test)
```

```python
print(classification_report(Y_test,Y_pred1))
print(confusion_matrix(Y_test,Y_pred1))
```

```python
## Now we will improve above accuracy using polynomial kernel
polynomial=SVC(kernel='poly')
polynomial.fit(X_train,Y_train)
## Predction
Y_pred2=polynomial.predict(X_test)
print(classification_report(Y_test,Y_pred2))
print(confusion_matrix(Y_test,Y_pred2))
```

```python
## Now we will improve above accuracy using sigmoid kernel
sig=SVC(kernel='sigmoid')
sig.fit(X_train,Y_train)
## Predction
Y_pred3=sig.predict(X_test)
print(classification_report(Y_test,Y_pred3))
print(confusion_matrix(Y_test,Y_pred3))
```

```python
## instead of manually, we can also give it all the params to tune itself. Hyperparameter Tuning*
```

```python
from sklearn.model_selection import GridSearchCV
params_grid = {'kernel' : ['rbf','sigmoid','linear'],
              'gamma' : [1,0.1,0.01,0.001],
              'C' : [0.1,1,10,100,1000]}
```

```python
grid=GridSearchCV(SVC(),param_grid=params_grid,refit=True,cv=5,verbose=3)
```

```python
grid.fit(X_train,Y_train)
```

```python
grid.best_params_
```

```python
Y_pred4=grid.predict(X_test)
print(classification_report(Y_test,Y_pred4))
print(confusion_matrix(Y_test,Y_pred4))
```
